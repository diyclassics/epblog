## Setup: the Greek Tesserae corpus via LatinCy Readers

```python
# latincyreaders exposes the CLTK Greek Tesserae corpus (auto-downloads on first use).
# BASIC annotation level = tokenization + lemmatization + POS (needs a Greek model).
from collections import Counter

from latincyreaders import GreekTesseraeReader, AnnotationLevel

# NB methodological choice to foreground: which Greek model does the lemmatizing?
# The package default is now LatinCy's grc_dep_web_lg; the older demo used OdyCy
# (grc_odycy_joint_sm). Different models -> different lemmas -> a different hapax set.
G = GreekTesseraeReader(annotation_level=AnnotationLevel.BASIC)
```

```python
# The Iliad and Odyssey are split across homer.iliad.part.N / homer.odyssey.part.N (books 1-24).
homer_files = G.fileids(match='homer')
print(f'{len(homer_files)} Homer files')
homer_files[:4]
```

## Hapaxes by lemma

```python
# concordance(basis='lemma') -> {lemma: [citations]}. A hapax legomenon (strict sense:
# once across Iliad + Odyssey combined) is any lemma with exactly one citation.
conc = G.concordance(fileids=homer_files, basis='lemma')
latincy_hapaxes = {lemma for lemma, cites in conc.items() if len(cites) == 1}
print(f'{len(conc)} unique lemmas; {len(latincy_hapaxes)} LatinCy hapaxes')
```

## Compare against the AGLDT treebank hapaxes

```python
# The archive post wrote its treebank hapaxes here.
from pathlib import Path

treebank_path = Path('..') / 'homeric-hapaxes' / 'data' / 'homeric_hapaxes.txt'
treebank_hapaxes = {line.strip() for line in treebank_path.read_text().splitlines() if line.strip()}

shared      = latincy_hapaxes & treebank_hapaxes
only_latincy   = latincy_hapaxes - treebank_hapaxes
only_treebank  = treebank_hapaxes - latincy_hapaxes

print(f'treebank hapaxes : {len(treebank_hapaxes)}')
print(f'latincy  hapaxes : {len(latincy_hapaxes)}')
print(f'shared           : {len(shared)}')
print(f'only in latincy  : {len(only_latincy)}')
print(f'only in treebank : {len(only_treebank)}')
```

```python
# Where they disagree is the interesting part: lemmatizer choices, edition/corpus
# differences (Tesserae text vs AGLDT), proper-noun handling, enclitics, etc.
from random import sample
print('sample only-in-latincy :', sample(sorted(only_latincy), min(10, len(only_latincy))))
print('sample only-in-treebank:', sample(sorted(only_treebank), min(10, len(only_treebank))))
```
