The *[scriptio continua]* Wikipedia entry includes a manuscript image with text from Virgil's *Georgics*. Let's start this exercise in reading unspaced Latin by picking a spot in the middle of the image and figuring out where we are. There is an *o* roughly in the center of the page and reading both a bit to the left and a bit to the right we can extend this to "moxetfrumentislaboradditusmalaculmos." (Note that the last two "letters" are in fact combined into a ligature; cf. @brown_1993_guide p. 16.) For today's post, I will concentrate just on that inital *o* and the adjacent letters.

How do we as (human) Latin readers know where words begin and end in this unspaced environment? We bring to the text knowledge that we already have about Latin orthography and vocabulary, among other things. We know from reading other Latin texts—likely in edited modern editions with proper spacing—that certain letter combinations are more common than others and, for this exercise in particular, we know that some letters and clusters of letters are more likely to appear at the beginnings of words and other at the ends. For example, *nt* appears frequently at the end of words, specifically third-person plural verbs, as well as in the middle of words (e.g. *frumentis*), but never at the beginning of a word. The letter *q* starts many Latin words—think about all of the interrogative and relative pronouns—but never at the end. We build up a mental model of Latin letter arrangement as we read and it only improves as we read more Latin.

We similarly develop instincts about word frequencies. We know that words *et* appear often and that words like *mox* and *labor* are common enough and that words like *frumentis* and *culmos* are relatively infrequent. With this lexicon of valid Latin words and rough estimations how often each word is likely to appear, we can make educated guesses when we enter an unspaced text about where to place word divisions.

Now, the letter *o* is a valid word of reasonably high frequency, usually introducing a vocative or standing as an interjection. As such, we cannot rule it out immediately as a self-contained lexical unit here. Of course, if we entertain this hypothesis, we are left with "...ab o radditus...". The form \**Radditus* is plausible enough, even if it is not really a Latin word; any word ending in *b* should give the Latin reader pause.

Let's look at just the words *labor additus* briefly to observe all of the possible splits, assuming here that we only have two words.

```python
# print all splits, cf. Let's look at just the words *labor additus* briefly to observe all of the possible splits, assuming here that we only have two words.

text = "laboradditus"

for i in range(1, len(text)):
    split = text[:i] + " " + text[i:]
    print(f'{i}: {split}')
```

In several of our enumerated splits, we see valid Latin words: *labora* in Split 6, *itus* in Split 8. We can make a case for *L* in Split 1 as a valid Roman numeral. But it only one of the eleven splits do we see two valid Latin words.

```python
# Load Latin word list from latincy-words repo
from pathlib import Path

LATINCY_WORDS = Path("/Volumes/fiona/work/code/diy/latincy-v3/latincy-words/data/final/latin-words.txt")

with open(LATINCY_WORDS) as f:
    latin_words = set(line.strip() for line in f if line.strip())

# The list stores proper nouns capitalized (Tullo, Numa, Roma), so lower-casing the
# *token* would miss them. Fold the list instead, and test through is_word().
latin_words_ci = {w.lower() for w in latin_words}


def is_word(form):
    """Case-insensitive membership test against latincy-words."""
    return form.lower() in latin_words_ci


print(f"Loaded {len(latin_words):,} Latin word forms ({len(latin_words_ci):,} case-folded)")
```

```python
# Let's make a table that shows split 1, split 2, split 1 in vocab, split 2 in vocab
from tabulate import tabulate

data = [
    [split, w1, w2, is_word(w1), is_word(w2), is_word(w1) and is_word(w2)]
    for i in range(1, len(text))
    for split in [text[:i] + " " + text[i:]]
    for w1, w2 in [split.split()]
]

print(tabulate(data, headers=["Split", "Word 1", "Word 2", "Word 1 in vocab", "Word 2 in vocab", "Both in vocab"]))
```

## A detailed example: *essetullo* from Livy 1.32

From the list above, let's examine one case in detail.

```python
# Livy, AUC 1.32: ... temporaque esse Tullo regi aptiora quam Numae.
text = "essetullo"

data = [
    [split, w1, w2, is_word(w1), is_word(w2), is_word(w1) and is_word(w2)]
    for i in range(1, len(text))
    for split in [text[:i] + " " + text[i:]]
    for w1, w2 in [split.split()]
]

print(f"All possible two-word splits of '{text}':\n")
print(tabulate(data, headers=["Split", "Word 1", "Word 2", "Word 1 in vocab", "Word 2 in vocab", "Both in vocab"]))

print("\n\nValid two-word splits (both words in vocabulary):")
for row in data:
    if row[5]:
        print(f"  {row[0]}")
```
